# Credit Risk Model

Проект по бинарной классификации клиентов для оценки риска дефолта по кредиту.

## Бизнес-цель
Снижение уровня просроченной задолженности банка путем автоматической оценки кредитоспособности заемщиков.

## Решение
Использован алгоритм **Random Forest Classifier** с применением:
- **`Pipeline`:** Для инкапсуляции шагов масштабирования (`StandardScaler`) и моделирования, предотвращая утечку данных.
- **`class_weight='balanced'`:** Механизм для компенсации сильного дисбаланса классов (дефолтов значительно меньше, чем не-дефолтов), чтобы модель уделяла внимание редким случаям.
- **`stratify=y`:** Гарантия сохранения пропорций классов в обучающей и тестовой выборках при разделении данных.

## Данные
Используется набор данных `credit_data.csv` с признаками: `age`, `income`, `debt_ratio`, `credit_score`, `number_of_open_loans`, `has_overdue`, и целевой переменной `default`.

## Результаты
- **ROC-AUC Score:** ~1.00 (значение может варьироваться из-за малого объема данных)
- **Отчет о классификации:** Предоставляет метрики Precision, Recall, F1-score для каждого класса, показывая, насколько хорошо модель различает "хороших" и "плохих" клиентов.
- **Важность признаков:** Отображает, какие факторы (например, `credit_score`, `debt_ratio`) наиболее сильно влияют на предсказание дефолта.

## Выводы
Модель способна ранжировать клиентов по вероятности дефолта, позволяя банку более эффективно управлять кредитными рисками.
