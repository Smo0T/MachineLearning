# Real Estate Price Estimator

Проект для автоматизированной оценки рыночной стоимости жилой недвижимости.

## Бизнес-цель
Предоставление быстрой и обоснованной оценки стоимости квартиры для использования в ипотечном кредитовании, анализе рынка или рекомендательных сервисах.

## Решение
Использован алгоритм **Random Forest Regressor** для предсказания цены.
- **Предобработка данных:**
    - Удаление выбросов: квартиры, чья цена попадает в верхние 5% (для предотвращения искажения модели).
    - Разделение данных на обучающую и тестовую выборки (`test_size=0.2`, `random_state=42`).
- **Обучение модели:** Модель обучается на признаках (`area`, `rooms`, `floor`, `distance_to_center`) для предсказания `price`.

## Данные
Используется набор данных `apartments.csv` с характеристиками квартир.

## Результаты
- **Средняя абсолютная ошибка (MAE):** ~847,600 руб. (показывает среднюю ошибку предсказания цены в абсолютном выражении).
- **Коэффициент детерминации (R²):** ~0.87 (показывает, что модель объясняет ~93% вариативности цен, что является отличным результатом для данного датасета).

## Выводы
Модель демонстрирует высокую точность в предсказании стоимости квартир на основе их основных характеристик. EDA показала, что `area` и `distance_to_center` являются ключевыми факторами, влияющими на цену.
