# Fraud Detection System

Система для автоматического выявления подозрительных (аномальных) транзакций.

## Бизнес-цель
Снижение финансовых потерь от мошенничества и ошибок путем оперативного выявления необычных банковских операций.

## Решение
Использован алгоритм **Isolation Forest**. Это метод машинного обучения без учителя (unsupervised learning), который эффективно выделяет аномалии (выбросы) в данных.
- **Признаки:** `amount` (сумма транзакции), `hour` (время суток).
- **Параметр `contamination`:** Установлен на `0.2` (20%), что означает, что модель ожидает, что примерно 20% транзакций являются аномальными. Это позволяет адаптировать чувствительность системы.

## Данные
Используется синтетический набор данных, смоделированный для демонстрации работы алгоритма. В реальных условиях используются реальные транзакционные данные.

## Результаты
- Алгоритм успешно выделяет транзакции с аномально высокой суммой и/или совершенные в необычное время суток (например, ночью).
- Выводятся список подозрительных транзакций (отмеченных как `-1`) и пример обычных транзакций (отмеченных как `1`).

## Выводы
Isolation Forest является мощным инструментом для обнаружения аномалий, позволяя банкам быстро реагировать на потенциальные мошеннические действия или системные сбои.
